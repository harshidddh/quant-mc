```python
import sys
sys.path.append("../src")

import numpy as np
import matplotlib.pyplot as plt
from quantlab.mc import simulate_gbm_paths

rng = np.random.default_rng(42)
T, n_steps = 1.0, 252
paths = simulate_gbm_paths(100, 0.05, 0.2, T, n_steps, 100, rng)
t = np.linspace(0, T, n_steps + 1)

plt.plot(t, paths.T, lw=0.8, alpha=0.7)
plt.plot(t, 100 * np.exp(0.05 * t), "k--", lw=2, label="expected price S0·e^(rt)")
plt.xlabel("time (years)")
plt.ylabel("stock price")
plt.title("100 simulated GBM paths")
plt.legend()
plt.show()
```
