```python
import sys
sys.path.append("../src")

import numpy as np
import matplotlib.pyplot as plt
from quantlab.bs import bs_call
from quantlab.mc import mc_european_call

S0, K, r, sigma, T = 100, 100, 0.05, 0.2, 1.0
exact = bs_call(S0, K, r, sigma, T)

Ns = np.logspace(3, 6, 7).astype(int)
n_seeds = 50

rmse = []
mean_se = []
for n in Ns:
    prices, ses = [], []
    for seed in range(n_seeds):
        p, se, _ = mc_european_call(S0, K, r, sigma, T, n, seed=seed)
        prices.append(p)
        ses.append(se)
    errors = np.array(prices) - exact
    rmse.append(np.sqrt(np.mean(errors**2)))
    mean_se.append(np.mean(ses))

slope = np.polyfit(np.log(Ns), np.log(rmse), 1)[0]
print("slope:", slope)

plt.loglog(Ns, rmse, "o-", label="actual error (RMSE over 50 seeds)")
plt.loglog(Ns, mean_se, "s--", label="reported standard error")
plt.xlabel("number of paths N")
plt.ylabel("error")
plt.legend()
plt.show()
```
